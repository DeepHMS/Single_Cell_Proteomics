```python
from google.colab import drive
import pandas as pd

# Step 1: Mount Google Drive
drive.mount('/content/drive')
```

```python
# Step 2: Provide the file path (update it with the correct path to your file)
file_path = '/content/drive/MyDrive/demo_.tsv'

# Step 3: Read the TSV file into a DataFrame
df = pd.read_csv(file_path, sep='\t')

# Step 4: Replace blanks with 0
df = df.fillna(0)

# Step 5: Save the output as a CSV file
output_path = 'Bulk_Lysate_Peptide_Output.csv'
df.to_csv(output_path, index=False)

# Step 6: Download the CSV file (optional)
from google.colab import files
files.download(output_path)

print("Processing complete. CSV file is ready for download.")
```
